0.96 → strong colour

-1 -------- 0 -------- +1
+1 → strong positive relationship
0 → no linear relationship
-1 → strong negative relationship

```python
import matplotlib.pyplot as plt

products=["Laptop","Mobile","TV","Watch","Tablet"]
sales=[120,180,90,140,80]

months=["Jan","Feb","Mar","Apr","May","Jun"]
revenue=[2.5,3.1,3.8,4.2,4.8,5.5]

expense=["Salary","Marketing","Rent","Transport"]
cost=[45,20,15,20]

salary=[25000,28000,30000,32000,35000,
        40000,42000,45000,47000,50000,
        52000,55000,60000]

plt.figure(figsize=(8,6))

plt.subplot(2,2,1)
plt.bar(products,sales)
plt.title("Product Sales")

plt.subplot(2,2,2)
plt.plot(months,revenue,marker="o")
plt.title("Revenue")

plt.subplot(2,2,3)
plt.pie(cost,labels=expense,autopct="%1.1f%%")
plt.title("Expense")

plt.subplot(2,2,4)
plt.hist(salary,bins=5)
plt.title("Salary Distribution")

plt.tight_layout()
plt.show()
```
